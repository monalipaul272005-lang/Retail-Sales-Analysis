```python

import pandas as pd

print("Retail Sales Analysis Project Started!")
```

```python

data = {
    "Order ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "City": [
        "Kolkata", "Mumbai", "Delhi", "Kolkata", "Mumbai",
        "Delhi", "Kolkata", "Mumbai", "Delhi", "Kolkata"
    ],
    "Product": [
        "Laptop", "Mobile", "Headphone", "Mobile", "Laptop",
        "Laptop", "Headphone", "Mobile", "Headphone", "Laptop"
    ],
    "Sales": [50000, 20000, 10000, 25000, 60000,
              40000, 15000, 30000, 12000, 70000],
    "Cost": [40000, 15000, 7000, 18000, 48000,
             32000, 10000, 22000, 8000, 55000]
}

df = pd.DataFrame(data)

df["Profit"] = df["Sales"] - df["Cost"]

df["Profit Margin (%)"] = (
    df["Profit"] / df["Sales"] * 100
).round(2)

print(df)
```

```python

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
```

```python

product_profit = df.groupby("Product")["Profit"].sum()

print(product_profit)

best_product = product_profit.idxmax()

print("Highest Profit Product:", best_product)
```

```python

city_sales = df.groupby("City")["Sales"].sum()

print(city_sales)

best_city = city_sales.idxmax()

print("Highest Sales City:", best_city)
```

```python

print("----- BUSINESS SUMMARY -----")
print(f"Total Sales: ₹{total_sales:,}")
print(f"Total Profit: ₹{total_profit:,}")
print(f"Highest Profit Product: {best_product}")
print(f"Highest Sales City: {best_city}")
```

```python

import matplotlib.pyplot as plt

# Product-wise sales analysis
product_sales = df.groupby("Product")["Sales"].sum()

# Create bar chart
product_sales.plot(kind="bar")

plt.title("Product-wise Total Sales")
plt.xlabel("Product")
plt.ylabel("Sales (₹)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt

product_profit = df.groupby("Product")["Profit"].sum()

product_profit.plot(kind="bar")

plt.title("Product-wise Total Profit")
plt.xlabel("Product")
plt.ylabel("Profit (₹)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

print("BUSINESS INSIGHTS")
print("------------------")

print("1. Laptop generates the highest total profit.")
print("2. Kolkata generates the highest total sales.")
print("3. The company should monitor product costs to improve profit.")
```

```python

print(df.head())
print(df.columns)
print(df.shape)
```

```python

# 1. Check dataset information
print("Dataset Information:")
df.info()

# 2. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 3. Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# 4. Check statistical summary
print("\nStatistical Summary:")
print(df.describe())
```

```python

# Check duplicate rows
print("Duplicate Rows:", df.duplicated().sum())

# Calculate Profit if Sales and Cost exist
df["Profit"] = df["Sales"] - df["Cost"]

# Product-wise sales and profit
summary = df.groupby("Product").agg(
    Total_Sales=("Sales", "sum"),
    Total_Cost=("Cost", "sum"),
    Total_Profit=("Profit", "sum")
)

print("\nProduct Performance:")
print(summary)
```

```python

# City-wise sales and profit
city_summary = df.groupby("City").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

print("City Performance Report")
print(city_summary)

# Find the city with highest sales
best_city = city_summary["Total_Sales"].idxmax()

print("\nHighest Sales City:", best_city)
```

```python

print("SALES STATISTICS REPORT")

print("Average Sales:", df["Sales"].mean())
print("Maximum Sales:", df["Sales"].max())
print("Minimum Sales:", df["Sales"].min())
print("Median Sales:", df["Sales"].median())
print("Standard Deviation:", df["Sales"].std())

print("\nAverage Profit:", df["Profit"].mean())
print("Total Profit Margin (%):",
      round(df["Profit"].sum() / df["Sales"].sum() * 100, 2))
```

```python

report = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Profit",
        "Average Sales",
        "Maximum Sales",
        "Total Profit Margin (%)"
    ],
    "Value": [
        df["Sales"].sum(),
        df["Profit"].sum(),
        df["Sales"].mean(),
        df["Sales"].max(),
        round(df["Profit"].sum() / df["Sales"].sum() * 100, 2)
    ]
})

print(report)

report.to_csv("retail_sales_report.csv", index=False)
```
